# Metrics to evaluate and test agents

* ## System metrics
* ## Quality metrics

## *system metrics (infrastructure perspective)*

1) **Latency (p50, p95, p99)**
    * How long users wait
        * Request timing in your application

2) **Error rate**
    * How often the agent fails entirely
        * Error counting in logs

3) **Tokens per task**
    * How much compute each task requires
        * LLM API response metadata

4) **Cost per task**
    * How much money each task costs
        * Token counts multiplied by pricing

5) **Tool call success rate**
    * How reliable external integrations are
        * Tool wrapper instrumentation

6) **Throughput**
    * How many requests the system handles	
        * Request counting


## *quality metrics (user’s perspective)*

1) **Correctness**	
    * Is the answer right?	
        * Ground truth comparison, human evaluation
2) **Trajectory quality**	
    * Did the agent take a reasonable path?	
        * Trajectory evaluation (see below)

3) **Helpfulness**	
    * Did the user get what they needed?	
        * User feedback, LLM-as-a-Judge
4) **Safety compliance**
    * Did the agent stay within bounds?	
        * Red-team testing, policy checkers

5) **Groundedness**	
    * Is the answer supported by evidence?	
        * Source attribution checking

6) **Coherence**
    * Does the response make sense?	
        * LLM-as-a-Judge, human evaluation
