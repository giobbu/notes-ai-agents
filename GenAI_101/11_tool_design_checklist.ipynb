## Tool design checklist

* [ 1 ] Name clearly describes the action (verb + noun)

* [ 2 ] Description explains what, when, and what it returns

* [ 3 ] Each tool has a single responsibility

* [ 4 ] Parameters have descriptive names (not abbreviations)

* [ 5 ] Parameters include descriptions with format examples

* [ 6 ] Required vs optional parameters are clearly marked

* [ 7 ] Output is concise - only fields the agent needs

* [ 8 ] Error messages are descriptive and actionable

* [ 9 ] Timeouts are implemented for external calls

* [ 10 ] Authentication is handled in the tool code, not by the model

* [ 11 ] Output size is bounded (truncation, pagination, or summarization)

* [ 12 ] Tool is tested independently before connecting to the agent
