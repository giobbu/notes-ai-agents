## Tool design checklist

* [ 1 ] Name clearly describes the action (verb + noun)

* [ 2 ] Description explains what, when, and what it returns

* [ 3 ] Each tool has a single responsibility

* [ 4 ] Parameters have descriptive names (not abbreviations)

* [ 5 ] Parameters include descriptions with format examples

* [ 6 ] Required vs optional parameters are clearly marked

* [ 7 ] Output is concise - only fields the agent needs

* [ 8 ] Error messages are descriptive and actionable

* [ 9 ] Timeouts are implemented for external calls

* [ 10 ] Authentication is handled in the tool code, not by the model

* [ 11 ] Output size is bounded (truncation, pagination, or summarization)

* [ 12 ] Tool is tested independently before connecting to the agent

## Common pitfalls

1) Too many tools overwhelm the context:

    * **Decision paralysis:** The model struggles to pick the right tool

    * **Wrong tool selection:** With many similar tools, the model picks the wrong one
    
    * **Wasted context:** Tool definitions crowd out useful conversation history

2) Vague tool description

3) Thin API wrappers (leaking implementation details)

4) Missing error information

5) Returning too much data:

    * **Filter at the source:** Only query what you need from the database/API

    * **Select relevant fields:** Do not return every field - pick what the agent needs
    
    * **Paginate:** Return a subset with the option to get more
    
    * **Summarize:** For large text responses, summarize before returning
    
    * **Truncate:** Cap long string fields (e.g., message[:200])
